```python
import re

def clean_text(text):
    """
    Clean extracted PDF text while preserving
    the original meaning and paragraph structure.
    """

    # Replace multiple spaces with a single space
    text = re.sub(r"[ \t]+", " ", text)

    # Replace excessive newlines with two newlines
    text = re.sub(r"\n{3,}", "\n\n", text)

    # Remove spaces at the beginning/end
    text = text.strip()

    return text


# Apply cleaning to every page
for document in all_documents:
    document.page_content = clean_text(document.page_content)

print(f"✅ Cleaned {len(all_documents)} document pages")
```

```python
def ask_research_bot(question, k=4):
    # Retrieve relevant chunks
    retrieved_docs = vector_db.as_retriever(
        search_type="mmr",
        search_kwargs={
            "k": k,
            "fetch_k": 12
        }
    ).invoke(question)

    # Combine retrieved research paper content
    context_parts = []

    for i, doc in enumerate(retrieved_docs, start=1):
        source = (
            f"[Source {i}: "
            f"{doc.metadata['paper_title']} "
            f"(Page {doc.metadata['page_number']})]"
        )

        context_parts.append(
            f"{source}\n{doc.page_content}"
        )

    context = "\n\n---\n\n".join(context_parts)

    # Create grounded prompt
    prompt = f"""
You are a Research Paper Answer Bot.

Answer the user's question using ONLY the research paper context below.

Rules:
1. Do not use outside knowledge.
2. If the answer is not available in the context, say:
   "I could not find enough information in the provided research papers."
3. Give a clear and concise answer.
4. Do not invent facts.
5. At the end, include the sources you used.

Research Paper Context:
{context}

User Question:
{question}

Answer:
"""

    # Generate answer
    response = llm.invoke(prompt)

    # Print answer
    print("=" * 70)
    print("ANSWER")
    print("=" * 70)
    print(response.content)

    print("\n" + "=" * 70)
    print("RETRIEVED SOURCES")
    print("=" * 70)

    # Show unique sources
    unique_sources = []

    for doc in retrieved_docs:
        source = (
            f"{doc.metadata['paper_title']} — "
            f"Page {doc.metadata['page_number']}"
        )

        if source not in unique_sources:
            unique_sources.append(source)

    for i, source in enumerate(unique_sources, start=1):
        print(f"{i}. {source}")

    return response.content, retrieved_docs
```

```python
def ask_research_bot(question, k=4):
    # Retrieve relevant chunks
    retrieved_docs = vector_db.as_retriever(
        search_type="mmr",
        search_kwargs={
            "k": k,
            "fetch_k": 12
        }
    ).invoke(question)

    # Build context with clear citation IDs
    context_parts = []
    source_map = {}

    for i, doc in enumerate(retrieved_docs, start=1):
        citation = (
            f"[{i}] {doc.metadata['paper_title']} "
            f"(Page {doc.metadata['page_number']})"
        )

        source_map[i] = citation

        context_parts.append(
            f"CITATION: {citation}\n"
            f"CONTENT:\n{doc.page_content}"
        )

    context = "\n\n---\n\n".join(context_parts)

    prompt = f"""
You are a Research Paper Answer Bot.

Answer the user's question using ONLY the provided research paper context.

STRICT RULES:
1. Do not use outside knowledge.
2. Do not invent information.
3. If the answer cannot be found in the context, say:
   "I could not find enough information in the provided research papers."
4. Write a clear and well-structured answer.
5. Add inline citations using ONLY the citation IDs provided in the context.
6. Every important factual statement should have a citation.

RESEARCH PAPER CONTEXT:
{context}

USER QUESTION:
{question}

ANSWER:
"""

    response = llm.invoke(prompt)

    print("=" * 70)
    print("ANSWER")
    print("=" * 70)
    print(response.content)

    print("\n" + "=" * 70)
    print("SOURCES")
    print("=" * 70)

    for citation_id, citation in source_map.items():
        print(citation)

    return response.content, retrieved_docs
```

```python
# ============================================================
# FINAL RAG ANSWER GENERATION
# ============================================================

import ollama


def generate_final_rag_answer(question):

    # --------------------------------------------------------
    # STEP 1: RETRIEVE RELEVANT DOCUMENTS
    # --------------------------------------------------------

    retrieved_docs = small_chunk_db.similarity_search(
        question,
        k=FINAL_TOP_K
    )

    # --------------------------------------------------------
    # STEP 2: BUILD CONTEXT
    # --------------------------------------------------------

    context = ""

    for i, doc in enumerate(retrieved_docs, start=1):

        paper = doc.metadata.get(
            "paper_title",
            "Unknown Paper"
        )

        page = doc.metadata.get(
            "page_number",
            doc.metadata.get("page", "Unknown")
        )

        context += f"""

SOURCE {i}
Paper: {paper}
Page: {page}

Content:
{doc.page_content}

--------------------------------------------------
"""

    # --------------------------------------------------------
    # STEP 3: CREATE RAG PROMPT
    # --------------------------------------------------------

    prompt = f"""
You are a research paper question answering assistant.

Answer the user's question using ONLY the information
provided in the retrieved research paper context.

Rules:

1. Do not use outside knowledge.
2. If the answer is not available in the context, say:
   "I could not find the answer in the available research papers."
3. Give a clear and concise answer.
4. At the end, mention the source number(s) used.

RETRIEVED RESEARCH PAPER CONTEXT:

{context}

USER QUESTION:

{question}

ANSWER:
"""

    # --------------------------------------------------------
    # STEP 4: GENERATE ANSWER USING OLLAMA
    # --------------------------------------------------------

    response = ollama.generate(
        model="llama3.2:latest",
        prompt=prompt
    )

    return response["response"], retrieved_docs


print("✅ Final RAG answer-generation function created successfully!")
```

```python
# ============================================================
# IMPROVED FINAL RAG ANSWER GENERATION
# ============================================================

def generate_improved_rag_answer(question):

    # --------------------------------------------------------
    # STEP 1: RETRIEVE DOCUMENTS
    # --------------------------------------------------------

    retrieved_docs = small_chunk_db.similarity_search(
        question,
        k=FINAL_TOP_K
    )

    # --------------------------------------------------------
    # STEP 2: BUILD CONTEXT
    # --------------------------------------------------------

    context = ""

    for i, doc in enumerate(retrieved_docs, start=1):

        paper = doc.metadata.get(
            "paper_title",
            "Unknown Paper"
        )

        page = doc.metadata.get(
            "page_number",
            doc.metadata.get("page", "Unknown")
        )

        context += f"""

SOURCE {i}
Paper: {paper}
Page: {page}

Content:
{doc.page_content}

--------------------------------------------------
"""

    # --------------------------------------------------------
    # STEP 3: IMPROVED PROMPT
    # --------------------------------------------------------

    prompt = f"""
You are a research paper question-answering assistant.

Answer the user's question using ONLY the retrieved
research paper context below.

IMPORTANT INSTRUCTIONS:

1. Carefully examine ALL retrieved sources before deciding
   that the answer is unavailable.

2. If one or more sources contain information that directly
   or indirectly answers the question, synthesize that
   information into a clear answer.

3. Do not simply repeat sentences from the papers.
   Explain the answer clearly and concisely using your own words.

4. Do not use outside knowledge.

5. Only say:
   "I could not find the answer in the available research papers."
   if NONE of the retrieved sources contain enough information
   to answer the question.

6. Ignore irrelevant retrieved sources.

7. When answering technical questions:
   - First provide a direct answer.
   - Then briefly explain how the concept works or why it matters,
     if the retrieved context supports it.

8. At the end, list the source number(s) actually used.

RETRIEVED RESEARCH PAPER CONTEXT:

{context}

USER QUESTION:

{question}

ANSWER:
"""

    # --------------------------------------------------------
    # STEP 4: GENERATE ANSWER
    # --------------------------------------------------------

    response = ollama.generate(
        model="llama3.2:latest",
        prompt=prompt
    )

    return response["response"], retrieved_docs


print("✅ Improved RAG answer-generation function created successfully!")
```

```python
# ============================================================
# STRICTER FINAL RAG FUNCTION
# ============================================================

def generate_strict_rag_answer(question):

    # Retrieve only the 3 most relevant documents
    results = small_chunk_db.similarity_search_with_score(
        question,
        k=3
    )

    retrieved_docs = [doc for doc, score in results]

    # Build context
    context = ""

    for i, doc in enumerate(retrieved_docs, start=1):

        paper = doc.metadata.get(
            "paper_title",
            "Unknown Paper"
        )

        page = doc.metadata.get(
            "page_number",
            doc.metadata.get("page", "Unknown")
        )

        context += f"""

SOURCE {i}
Paper: {paper}
Page: {page}

Content:
{doc.page_content}

--------------------------------------------------
"""

    # Strict grounding prompt
    prompt = f"""
You are a research paper question-answering assistant.

Answer the question using ONLY information explicitly
supported by the retrieved context.

STRICT RULES:

1. Use only the retrieved sources below.

2. Do NOT use general knowledge.

3. Do NOT infer technical mechanisms that are not explicitly
   described in the retrieved text.

4. Do NOT combine unrelated information from different papers
   into a single answer.

5. Ignore information that does not directly answer the question.

6. If the retrieved context provides only a partial answer,
   clearly state that the available research papers provide
   only a partial answer.

7. If the answer is not supported by the retrieved context, say:
   "I could not find a supported answer in the retrieved research papers."

8. Give the direct answer first.

9. End with:
   Sources used: SOURCE X, SOURCE Y

RETRIEVED CONTEXT:

{context}

QUESTION:

{question}

ANSWER:
"""

    response = ollama.generate(
        model="llama3.2:latest",
        prompt=prompt
    )

    return response["response"], retrieved_docs


print("✅ Strict RAG function created successfully!")
```
